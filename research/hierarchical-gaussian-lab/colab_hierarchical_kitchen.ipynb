# Real kitchen: incremental anisotropic Gaussian mapping and hierarchical semantics

279 real Mip-NeRF 360 kitchen photographs of an object-centered tabletop orbit; offline COLMAP poses / triangulated coordinates. Prefix-restricted mapping, 300k / 600k caps, replay ablation, SAM 2.1 + scale-conditioned affinity + multi-view CLIP descriptors. Semantic training is currently a final-map post-pass. Fine masks are candidate parts; there is no human semantic GT. This is not a complete online SLAM. All source is embedded. Use an NVIDIA GPU.


```python
#@title Unpack complete source and check GPU
from pathlib import Path
import io,zipfile,base64,os,sys,subprocess,json
root=Path('/content/hierarchical-gaussian'); root.mkdir(exist_ok=True)
zipfile.ZipFile(io.BytesIO(base64.b64decode(''))).extractall(root)
os.chdir(root); os.environ['MAX_JOBS']='2'; os.environ['OPENBLAS_NUM_THREADS']='2'
import torch
assert torch.cuda.is_available(), 'Select a GPU runtime'
print(torch.cuda.get_device_name(), root)
def run(*args):
    proc=subprocess.Popen([sys.executable,*args],stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True)
    for line in proc.stdout: print(line,end='',flush=True)
    if proc.wait(): raise RuntimeError('Command failed: '+str(args))
```

```python
#@title Install official CUDA and segmentation dependencies
run('-m','pip','install','-q','gsplat==1.5.3','scipy','scikit-image','matplotlib','open_clip_torch','remotezip')
os.environ['SAM2_BUILD_CUDA']='0'
run('-m','pip','install','-q','--no-build-isolation','git+https://github.com/facebookresearch/sam2.git@2b90b9f5ceec907a1c18123530e92e794ad901a4')
run('-m','unittest','discover','-s','tests','-v')
```

```python
#@title Download the real kitchen scene from the official release
from remotezip import RemoteZip
with RemoteZip('https://storage.googleapis.com/gresearch/refraw360/360_v2.zip') as z:
    fs=[f for f in z.infolist() if f.filename.startswith('kitchen/') and ('/images_4/' in f.filename or '/sparse/' in f.filename)]
    print('Files:',len(fs),'compressed MB:',sum(f.compress_size for f in fs)/1e6)
    for f in fs: z.extract(f,'/content/mip360')
from hglab.colmap import read_scene
scene=read_scene('/content/mip360/kitchen',800)
print('Actual photographs:',len(scene['images']))
```

```python
#@title Train independent capacity / old-region replay variants
for name,cap,replay in [('kitchen-300k-l1',300000,.3),('kitchen-600k-l1',600000,.3),('kitchen-600k-no-replay',600000,0.)]:
    run('-u','-m','hglab.train','--data','/content/mip360/kitchen','--output','results/'+name,'--cap',str(cap),'--replay',str(replay),'--width','800','--stages','6','--steps','1800')
```

```python
#@title Predict nested masks and learn scale-conditioned semantic affinity
run('-u','-m','hglab.extract_masks','--data','/content/mip360/kitchen','--output','semantics','--width','800','--every','6')
run('-u','-m','hglab.semantic_field','--checkpoint','results/kitchen-300k-l1/checkpoint.pt','--masks','semantics','--output','semantic-field-300k','--steps','1200')
```

```python
#@title Export real renders, measurements, maps and local replay
run('-u','-m','hglab.bundle')
import pandas as pd
from IPython.display import display
from PIL import Image
display(pd.DataFrame(json.loads(Path('deliverable/comparison.json').read_text())).round(3))
print(json.dumps(json.loads(Path('semantic-field-300k/metrics.json').read_text())['summary'],indent=2))
display(Image.open('deliverable/assets/comparison-136.jpg'))
from google.colab import files
files.download('/content/hierarchical-kitchen-results.zip')
```
