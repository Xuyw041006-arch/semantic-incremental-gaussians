# Semantic incremental Gaussian: full S/B/R factorial

Two real scenes, eight variants, three paired seeds. Known poses; SAM pseudo-regions. No strict online RGB-SLAM or human part-accuracy claim. This notebook regenerates caches unless archived inputs are restored.


```python
from pathlib import Path
import io,zipfile,base64,os,sys,subprocess,json
root=Path('/content/semantic-incremental');root.mkdir(exist_ok=True)
zipfile.ZipFile(io.BytesIO(base64.b64decode(''))).extractall(root)
os.chdir(root)
os.environ['MAX_JOBS']='2';os.environ['SAM2_BUILD_CUDA']='0'
def run(*args):
 subprocess.run([sys.executable,*args],check=True)
run('bootstrap_colab.py')
```

```python
# Builds posed RGB-D manifest, shared SAM/CLIP caches and prefix-only region graphs.
run('prepare_data.py')
print((root/'experiment-plan.json').read_text())
```

```python
# Independent subprocesses; randomized variant order inside dataset/seed blocks.
# Reuse only directories with the same protocol. Completed run.json groups are skipped.
# An unfinished group restarts from its first stage, not a checkpoint.
# Guard against duplicate GPU runs; archive stale completion/failure markers.
import psutil,time
for process in psutil.process_iter(['pid','cmdline','cwd']):
 try:
  if process.info['cwd']==str(root) and any('run_ablation.py'==Path(x).name for x in (process.info['cmdline'] or [])):
   raise RuntimeError('An ablation runner is already active; inspect progress instead.')
 except (psutil.NoSuchProcess,psutil.AccessDenied):pass
stamp=str(int(time.time()))
for name in ['status.json','failure.txt','finish-status.json','finish-failure.txt']:
 p=root/name
 if p.exists():p.rename(root/(name+'.previous-'+stamp))
(root/'status.json').write_text(json.dumps(dict(state='starting',completed=[],total=48)))
job=subprocess.Popen([sys.executable,'-u','run_ablation.py'],cwd=root,stdout=open('/content/semantic-ablation.log','w'),stderr=subprocess.STDOUT,start_new_session=True)
monitor=subprocess.Popen([sys.executable,'monitor_memory.py'],cwd=root,stdout=open('/content/semantic-monitor.log','w'),stderr=subprocess.STDOUT,start_new_session=True)
finish=subprocess.Popen([sys.executable,'-u','finish_experiment.py'],cwd=root,stdout=open('/content/semantic-finish.log','w'),stderr=subprocess.STDOUT,start_new_session=True)
print('runner, monitor, audit manager:',job.pid,monitor.pid,finish.pid)
```

```python
# Rerun to inspect progress.
for name in ['status.json','failure.txt','finish-status.json','finish-failure.txt']:
 p=root/name
 if p.exists():print(name,p.read_text()[-3000:])
print(Path('/content/semantic-finish.log').read_text()[-3000:])
```
